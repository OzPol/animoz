````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.animation import FuncAnimation

# Parameters
r_tube = 0.1
num_steps = 36
step_size = 1 / num_steps

# Generate hypotrochoid trefoil knot points
def generate_knot():
    points = []
    for N in range(-1, num_steps + 2):
        theta = 2 * np.pi * N * step_size
        x = 0.3 * (2 * np.sin(2 * theta) - np.sin(theta))
        y = 0.3 * (2 * np.cos(2 * theta) + np.cos(theta))
        z = 0.3 * np.sin(3 * theta)
        points.append((x, y, z))
    return np.array(points)

points = generate_knot()

# Set up the 3D plot
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
line, = ax.plot([], [], [], lw=2, color=(0, 0.25, 1))

def init():
    ax.set_xlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.set_zlim(-1, 1)
    ax.set_facecolor('white')
    return line,

def update(frame):
    angle = frame
    R = np.array([  [np.cos(np.radians(angle)), 0, np.sin(np.radians(angle))],
                    [0, 1, 0],
                    [-np.sin(np.radians(angle)), 0, np.cos(np.radians(angle))]])
    rotated = points @ R.T
    line.set_data(rotated[:, 0], rotated[:, 1])
    line.set_3d_properties(rotated[:, 2])
    return line,

anim_n = FuncAnimation(fig, update, init_func=init, frames=360, interval=40, blit=False)
# anim_n = FuncAnimation(fig, update, frames=360, interval=30, blit=False)
anim_n.save("output/00_Test/trefoil_knot_rotating_n.gif", writer="pillow", fps=30)
# plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Tube parameters
r = 0.1
t_vals = np.linspace(0, 2*np.pi, 500)
theta_vals = np.linspace(0, 2*np.pi, 20)

# Knot centerline and derivatives
def trefoil_knot(t):
    x = np.sin(t) + 2*np.sin(2*t)
    y = np.cos(t) - 2*np.cos(2*t)
    z = -np.sin(3*t)
    return x, y, z

def trefoil_tangent(t):
    dx = np.cos(t) + 4*np.cos(2*t)
    dy = -np.sin(t) + 4*np.sin(2*t)
    dz = -3*np.cos(3*t)
    return dx, dy, dz

# Normalize vectors
def normalize(x, y, z):
    norm = np.sqrt(x**2 + y**2 + z**2)
    return x/norm, y/norm, z/norm

# Build tube coordinates
def build_tube():
    xk, yk, zk = trefoil_knot(t_vals)
    dx, dy, dz = trefoil_tangent(t_vals)
    tx, ty, tz = normalize(dx, dy, dz)

    # Arbitrary perpendicular vector
    bx, by, bz = normalize(-ty, tx, 0)
    nx, ny, nz = np.cross([tx, ty, tz], [bx, by, bz], axis=0)

    T, Theta = np.meshgrid(t_vals, theta_vals, indexing='ij')
    X = xk[:, None] + r * (bx[:, None] * np.cos(Theta) + nx[:, None] * np.sin(Theta))
    Y = yk[:, None] + r * (by[:, None] * np.cos(Theta) + ny[:, None] * np.sin(Theta))
    Z = zk[:, None] + r * (bz[:, None] * np.cos(Theta) + nz[:, None] * np.sin(Theta))
    return X, Y, Z

# Rotate geometry using Y-axis matrix
def rotate_y(X, Y, Z, angle_deg):
    angle_rad = np.radians(angle_deg)
    cos_a = np.cos(angle_rad)
    sin_a = np.sin(angle_rad)
    X_rot = X * cos_a + Z * sin_a
    Z_rot = -X * sin_a + Z * cos_a
    return X_rot, Y, Z_rot

# Initial surface
X_base, Y_base, Z_base = build_tube()

# Plot setup
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_zlim(-2, 2)
ax.axis('off')
surf = [ax.plot_surface(X_base, Y_base, Z_base, color=(0, 0.25, 1), linewidth=0, antialiased=False)]

# Animation
def update(frame):
    if surf[0] in ax.collections:
        surf[0].remove()
    X_rot, Y_rot, Z_rot = rotate_y(X_base, Y_base, Z_base, angle_deg=frame)
    surf[0] = ax.plot_surface(X_rot, Y_rot, Z_rot, color=(0, 0.25, 1), linewidth=0, antialiased=False)
    return surf

anim = FuncAnimation(fig, update, frames=360, interval=30, blit=False)
anim.save("output/00_Test/trefoil_knot_true_rotate.gif", writer="pillow", fps=30)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Tube parameters
r = 0.1
t_vals = np.linspace(0, 2*np.pi, 500)
theta_vals = np.linspace(0, 2*np.pi, 30)

# Knot centerline and derivatives
def trefoil_knot(t):
    x = np.sin(t) + 2*np.sin(2*t)
    y = np.cos(t) - 2*np.cos(2*t)
    z = -np.sin(3*t)
    return x, y, z

def trefoil_tangent(t):
    dx = np.cos(t) + 4*np.cos(2*t)
    dy = -np.sin(t) + 4*np.sin(2*t)
    dz = -3*np.cos(3*t)
    return dx, dy, dz

# Normalize vectors
def normalize(x, y, z):
    norm = np.sqrt(x**2 + y**2 + z**2)
    return x/norm, y/norm, z/norm

# Build tube coordinates
def build_tube():
    xk, yk, zk = trefoil_knot(t_vals)
    dx, dy, dz = trefoil_tangent(t_vals)
    tx, ty, tz = normalize(dx, dy, dz)

    # Arbitrary perpendicular vector
    bx, by, bz = normalize(-ty, tx, 0)
    nx, ny, nz = np.cross([tx, ty, tz], [bx, by, bz], axis=0)

    T, Theta = np.meshgrid(t_vals, theta_vals, indexing='ij')
    X = xk[:, None] + r * (bx[:, None] * np.cos(Theta) + nx[:, None] * np.sin(Theta))
    Y = yk[:, None] + r * (by[:, None] * np.cos(Theta) + ny[:, None] * np.sin(Theta))
    Z = zk[:, None] + r * (bz[:, None] * np.cos(Theta) + nz[:, None] * np.sin(Theta))
    return X, Y, Z

# Composite 3D rotation (Euler-style)
def rotate_composite(X, Y, Z, angle_deg):
    angle_rad = np.radians(angle_deg)
    cos_a, sin_a = np.cos(angle_rad), np.sin(angle_rad)

    # Z rotation
    x1 = X * cos_a - Y * sin_a
    y1 = X * sin_a + Y * cos_a
    z1 = Z

    # Y rotation
    x2 = x1 * cos_a + z1 * sin_a
    y2 = y1
    z2 = -x1 * sin_a + z1 * cos_a

    # Slight X rotation (tumble)
    y3 = y2 * cos_a - z2 * sin_a
    z3 = y2 * sin_a + z2 * cos_a
    x3 = x2

    return x3, y3, z3

# Build base tube
X_base, Y_base, Z_base = build_tube()

# Setup 3D figure
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_zlim(-2, 2)
ax.axis('off')
surf = [ax.plot_surface(X_base, Y_base, Z_base, color=(0, 0.25, 1), linewidth=0, antialiased=True)]

# Animation update
def update(frame):
    if surf[0] in ax.collections:
        surf[0].remove()
    Xr, Yr, Zr = rotate_composite(X_base, Y_base, Z_base, angle_deg=frame)
    surf[0] = ax.plot_surface(Xr, Yr, Zr, color=(0, 0.25, 1), linewidth=0, antialiased=True)
    return surf

# Animate
anim = FuncAnimation(fig, update, frames=180, interval=30, blit=False)
anim.save("output/00_Test/trefoil_knot_tumbling.gif", writer="pillow", fps=30)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Tube parameters
r = 0.2  # Increase to adjust the fatness of the knot 
t_vals = np.linspace(0, 2 * np.pi, 500)
theta_vals = np.linspace(0, 2 * np.pi, 30)

def torus_trefoil_knot(t):
    x = (2 + np.cos(3 * t)) * np.cos(2 * t)
    y = (2 + np.cos(3 * t)) * np.sin(2 * t)
    z = np.sin(3 * t)
    return x, y, z

def torus_trefoil_tangent(t):
    dx = -2 * (2 + np.cos(3 * t)) * np.sin(2 * t) - 3 * np.sin(3 * t) * np.cos(2 * t)
    dy =  2 * (2 + np.cos(3 * t)) * np.cos(2 * t) - 3 * np.sin(3 * t) * np.sin(2 * t)
    dz =  3 * np.cos(3 * t)
    return dx, dy, dz

def normalize(x, y, z):
    norm = np.sqrt(x ** 2 + y ** 2 + z ** 2)
    return x / norm, y / norm, z / norm

def build_tube():
    xk, yk, zk = torus_trefoil_knot(t_vals)
    dx, dy, dz = torus_trefoil_tangent(t_vals)
    tx, ty, tz = normalize(dx, dy, dz)
    bx, by, bz = normalize(-ty, tx, 0)
    nx, ny, nz = np.cross([tx, ty, tz], [bx, by, bz], axis=0)
    T, Theta = np.meshgrid(t_vals, theta_vals, indexing='ij')
    X = xk[:, None] + r * (bx[:, None] * np.cos(Theta) + nx[:, None] * np.sin(Theta))
    Y = yk[:, None] + r * (by[:, None] * np.cos(Theta) + ny[:, None] * np.sin(Theta))
    Z = zk[:, None] + r * (bz[:, None] * np.cos(Theta) + nz[:, None] * np.sin(Theta))
    return X, Y, Z

def rotate_y(X, Y, Z, angle_deg):
    angle_rad = np.radians(angle_deg)
    cos_a, sin_a = np.cos(angle_rad), np.sin(angle_rad)
    X_rot = X * cos_a + Z * sin_a
    Z_rot = -X * sin_a + Z * cos_a
    return X_rot, Y, Z_rot

X_base, Y_base, Z_base = build_tube()

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_zlim(-2, 2)
ax.axis('off')
surf = [ax.plot_surface(X_base, Y_base, Z_base, color=(0, 0.25, 1), linewidth=0, antialiased=True)]

def update(frame):
    angle = frame
    if surf[0] in ax.collections:
        surf[0].remove()
    Xr, Yr, Zr = rotate_y(X_base, Y_base, Z_base, angle)
    surf[0] = ax.plot_surface(Xr, Yr, Zr, color=(0, 0.25, 1), linewidth=0, antialiased=True)
    return surf

anim = FuncAnimation(fig, update, frames=360, interval=30, blit=False)
anim.save("output/00_Test/trefoil_torus_knot.gif", writer="pillow", fps=30)
````
